```python
import numpy as np
import matplotlib.pyplot as plt


# Tempo de 0 a 50 horas
t = np.linspace(0, 50, 300)

# Sinal principal (suave) simulando um padrão diário de NDVI
ndvi_predito = 0.5 + 0.1 * np.sin(2 * np.pi * t / 30)

# NDVI Real com ruído adicional
ruido = np.random.normal(0, 0.015, size=len(t))
ndvi_real = ndvi_predito + ruido

# --- Plotagem ---
plt.figure(figsize=(10, 5))
plt.plot(t, ndvi_real, label="NDVI Real", color="tab:blue")
plt.plot(t, ndvi_predito, label="NDVI Predito", color="orange", linewidth=2)

plt.xlabel("Tempo (h)")
plt.ylabel("NDVI")
plt.title("")
plt.grid(True, linestyle="--", alpha=0.4)

plt.legend()

# Ajuste de espaçamento
plt.tight_layout()

plt.show()
```
